```python
!pip install seaborn matplotlib pandas
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

**Dataset**

```python
Students = ['Amit','Riya','John','Sara','Ali','Neha','Raj','Meena']
Marks=[78,85,60,92,55,88,73,95]
Study_Hours =[4,5,2,6,3,5,4,6]
Attendance = [90,95,70,98,60,92,85,99]
Sports_Score=[30,30,20,50,25,45,35,55]
df = pd.DataFrame({
    'Students': Students,
    'Marks': Marks,
    'Study Hours': Study_Hours,
    'Attendance': Attendance,
    'Sports Score': Sports_Score
})
print(df)
```

**Bar chart for Marks**

```python
plt.bar(Students,Marks ,color=['lightblue','lavender'],width=0.6)
plt.title("Marks of Students")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.show()
```

**Line chart for Marks**

```python
plt.plot(Students,Marks, color="g",linewidth=2,marker="o",markersize=10, markerfacecolor="y",label="Marks",linestyle="--")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.title("Marks Of Students in Sequence")
plt.legend()
plt.show()
```

**Pie chart for Attendance**

```python
plt.figure(figsize=(8,8))
col =('forestgreen','limegreen',"olive",'sandybrown',"peru","skyblue","teal","gold")
plt.pie(Attendance, labels= Students,autopct='%1.1f%%',startangle=90,shadow=True, colors=col)
plt.title("Attendance distribution")
plt.show()
```

**Histogram of marks**

```python
plt.hist(Marks, bins=6,color='teal',edgecolor='black')
plt.title("Distribution Of Marks")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.show()
```

**Scatterplot : Study Hours vs Marks**

```python
plt.scatter(Study_Hours,Marks,color='limegreen',s=150,marker='*',edgecolor=("forestgreen"))
plt.title ("Study Hours vs Marks")
plt.xlabel("Study hours")
plt.ylabel("Marks")
plt.grid(True)
plt.show()
```

**Bar chart for Sports Scores**

```python
plt.bar(Students,Sports_Score ,color=['darkgreen','mediumseagreen'],width=0.6,edgecolor="forestgreen")
plt.title("sports scores of students")
plt.xlabel("Students")
plt.ylabel("Sports score")

plt.show()
```

**Scatterplot : Attendance vs Marks**

```python
plt.scatter(Attendance,Marks,color='gold',s=150,marker='p',edgecolor=("forestgreen"))
plt.title ("Attendance vs Marks")
plt.xlabel("Attendance")
plt.ylabel("Marks")
plt.grid(True)
plt.show()
```

**Comparison between Study hours and Marks**

```python
x= np.arange(len(Students))
width=0.25
bar1=plt.bar(x-width/2,Marks, width, label='Marks',color="brown",edgecolor="maroon") 
bar2=plt.bar(x+width/2,Study_Hours, width, label="Study_Hours",color="pink", edgecolor="maroon") 
plt.xticks(x, Students)
plt.bar_label(bar1)
plt.bar_label(bar2)
plt.legend()

plt.show()
```

**Histogram of study hours**

```python
plt.figure(figsize=(8,6))
plt.hist(Study_Hours, bins=6,color='grey',edgecolor='black')
plt.title("Distribution of study hours")
plt.xlabel("Study hours")
plt.ylabel("Frequency")
plt.show()
```

**Top 3 students by marks**

```python
top3 = df.sort_values(by='Marks',
                      ascending=False).head(3)

plt.figure(figsize=(6,5))
plt.bar(top3['Students'], top3['Marks'],color="indigo")
plt.title("Top 3 Students Based on Marks")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.show()

print("\nTop 3 Students")
print(top3[['Students', 'Marks']])
```

**Compare Marks and Sports Scores**

```python
plt.plot(Students,Marks,color="turquoise", linewidth=2.2,marker="p",markersize=10,markerfacecolor="indigo",label="Marks",linestyle="-.")
plt.plot(Students,Sports_Score,color="indigo", linewidth=2.2,marker="D",markersize=10,markerfacecolor="turquoise",label="Student_Score",linestyle="-.")
plt.xlabel("Students")
plt.ylabel("Sports Score")
plt.title("Marks vs Sports Scores")
plt.grid()
plt.legend()
plt.show()
```

**Relationship Between Study Hours and Marks**

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    x=Study_Hours,
    y=Marks, color= "brown", marker="*", s=150,markers=True, alpha=0.8,linewidth=2,edgecolor="black"
)

plt.title("Relationship Between Study Hours and Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")

plt.show()
```

**Relationship Between Attendance and Marks**


```python
plt.figure(figsize=(6,5))

sns.scatterplot(
    x=Attendance,
    y=Marks, color= "indigo", marker="1", s=150,markers=True, alpha=0.8,linewidth=2
)

plt.title("Relationship Between attendance and Marks")
plt.xlabel("Attendace")
plt.ylabel("Marks")

plt.show()
```
